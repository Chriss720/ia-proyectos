# Actividades: Transformers
Resolución de las actividades del manual de Transformers.

## Actividad 1: La matriz de atención

### Paso 1: Fila de COME (Puntuación de 0 a 10)

| | EL | GATO | COME | PESCADO |
| --- | :---: | :---: | :---: | :---: |
| **Desde COME →** | 2 | 8 | 10 | 9 |

### Paso 2 — Convertir a porcentajes (mini-softmax)
* **Suma de puntuaciones:** 29

| Palabra | Puntuación | ÷ Suma | × 100 ≈ % |
| --- | :---: | :---: | :---: |
| **EL** | 2 | 2/29 | 6.9% |
| **GATO** | 8 | 8/29 | 27.6% |
| **COME** | 10 | 10/29 | 34.5% |
| **PESCADO** | 9 | 9/29 | 31.0% |
| **Total** | **29** | - | **100%** |

### Paso 3 — Interpretación
* **¿A quién le diste más atención? ¿Tiene sentido para el verbo “come”?**
  * Le di más atención a **COME** (auto-atención) y a **PESCADO** (objeto directo), seguido por **GATO** (sujeto). Sí tiene sentido, porque para entender el verbo "come" necesitas saber qué se come (pescado) y quién realiza la acción (gato).

### Pregunta de cierre
* **Si fueras la palabra PESCADO, ¿crees que tu fila de porcentajes sería igual? ¿Por qué sí o por no?**
  * No. Para **PESCADO**, la palabra más importante sería **COME** (para saber qué acción le ocurre), mientras que **GATO** o **EL** tendrían una importancia directa casi nula. La atención cambia según la perspectiva.

## Actividad 2: La palabra ambigua (dos contextos)

### Frase A: FUIMOS AL BANCO DEL RIO
* **Puntuación y porcentajes desde BANCO:**
  | | FUIMOS | AL | BANCO | DEL | RIO |
  | --- | :---: | :---: | :---: | :---: | :---: |
  | **Puntos** | 1 | 1 | 10 | 6 | 9 |
  | **%** | 3.7% | 3.7% | 37.0% | 22.2% | 33.3% |

### Frase B: FUIMOS AL BANCO A SACAR DINERO
* **Puntuación y porcentajes desde BANCO:**
  | | FUIMOS | AL | BANCO | A | SACAR | DINERO |
  | --- | :---: | :---: | :---: | :---: | :---: | :---: |
  | **Puntos** | 1 | 1 | 10 | 1 | 8 | 9 |
  | **%** | 3.3% | 3.3% | 33.3% | 3.3% | 26.7% | 30.0% |

### Preguntas
1. **¿En cuál frase BANCO le da más puntos a “RIO” / “DEL”?**
   * En la **Frase A**.
2. **¿En cuál le da más a “DINERO” / “SACAR”?**
   * En la **Frase B**.
3. **Explicación (máximo 3 líneas):**
   * Esto imita a los Transformers porque la misma palabra cambia sus pesos de atención según los vecinos que la acompañan, adaptando su significado al contexto.

## Actividad 3: Máscara causal (no hacer trampa)

### Cuadrícula de atención (✓ = permitido mirar, ✗ = bloqueado)

| Palabra que genera \ Palabra mirada | EL | GATO | COME | PESCADO |
| --- | :---: | :---: | :---: | :---: |
| **1º EL** | ✓ | ✗ | ✗ | ✗ |
| **2º GATO** | ✓ | ✓ | ✗ | ✗ |
| **3º COME** | ✓ | ✓ | ✓ | ✗ |
| **4º PESCADO** | ✓ | ✓ | ✓ | ✓ |

### Preguntas
1. **¿Cuántos ✓ hay en la fila de la última palabra (PESCADO)?**
   * **4 ✓**.
2. **¿Cuántos ✓ hay en la fila de la primera palabra (EL)?**
   * **1 ✓**.
3. **¿Por qué creen que es necesaria para escribir texto?**
   * Porque al generar texto palabra por palabra, el modelo no debe conocer las palabras del futuro (ya que aún no existen). La máscara evita que haga trampa en el entrenamiento.

## Actividad 4: Varias cabezas (varios criterios)

Frase: **MARIA NO COMIO PORQUE ESTABA ENFERMA**

### Cabezas de atención para la palabra COMIO (Puntos y Porcentajes)

* **Cabeza A (Causa): ¿Quién explica el porqué?**
  | | MARIA | NO | COMIO | PORQUE | ESTABA | ENFERMA |
  | --- | :---: | :---: | :---: | :---: | :---: | :---: |
  | **Puntos** | 1 | 1 | 10 | 7 | 3 | 9 |
  | **%** | 3.2% | 3.2% | 32.3% | 22.6% | 9.7% | 29.0% |

* **Cabeza B (Sujeto): ¿Quién realiza la acción?**
  | | MARIA | NO | COMIO | PORQUE | ESTABA | ENFERMA |
  | --- | :---: | :---: | :---: | :---: | :---: | :---: |
  | **Puntos** | 10 | 1 | 10 | 1 | 1 | 1 |
  | **%** | 41.7% | 4.2% | 41.7% | 4.2% | 4.2% | 4.2% |

* **Cabeza C (Vecinos): Vecinos inmediatos**
  | | MARIA | NO | COMIO | PORQUE | ESTABA | ENFERMA |
  | --- | :---: | :---: | :---: | :---: | :---: | :---: |
  | **Puntos** | 1 | 10 | 10 | 10 | 1 | 1 |
  | **%** | 3.0% | 30.3% | 30.3% | 30.3% | 3.0% | 3.0% |

### Preguntas
1. **¿Las tres filas son iguales?**
   * No, cada una se enfoca en relaciones distintas (causa, sujeto o cercanía).
2. **¿Qué ventaja tendría ver la frase desde tres criterios y no solo uno?**
   * Permite al Transformer analizar múltiples aspectos gramaticales y de significado al mismo tiempo, obteniendo una comprensión mucho más completa.

## Actividad 5: Encoder vs Decoder (role-play)

### Respuestas del Debrief
1. **¿Qué fue más fácil: leer todo (encoder) o escribir de a poco (decoder)?**
   * Leer todo (encoder) es más fácil porque tienes toda la frase completa disponible. Escribir (decoder) es más difícil porque debes ir palabra por palabra cuidando el sentido sin ver el futuro.
2. **¿En qué momento el decoder “necesitó” mirar atrás?**
   * Al generar cada nueva palabra para mantener concordancia y coherencia gramatical (atención causal) y para conectar con las pistas del encoder (atención cruzada).
3. **¿Cómo se relaciona esto con traducir o con un chatbot?**
   * El encoder procesa el texto original entero (ej. una pregunta) y el decoder genera la respuesta o traducción palabra por palabra guiado por el análisis del encoder.
