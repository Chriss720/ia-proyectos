## 2. Actividad 2.1: Mapeo de Variables

Dada la ecuación fundamental de la Vanilla RNN: 

$$h_t = \tanh(W_{hx} x_t + W_{hh} h_{t-1} + b)$$

A continuación relacionamos cada variable y función con su estrofa correspondiente en el acertijo literario.

### La entrada actual ($x_t$)
La frase asociada es: *"Soy la novedad pura, el pulso del instante, la matriz de características que el mundo me da en este segundo."* Representa la información que entra a la red en el paso de tiempo actual, como por ejemplo la palabra actual en una frase o la lectura de un sensor en este instante.

### El estado oculto anterior ($h_{t-1}$)
La frase asociada es: *"el fantasma del pasado, que trae consigo el resumen de todo lo que hemos vivido hasta ayer."* Es la memoria acumulada que la red arrastra desde el inicio hasta el paso de tiempo anterior, sirviendo como contexto para interpretar la nueva entrada.

### Las matrices de pesos ($W_{hx}, W_{hh}$)
La frase asociada es: *"peajes inmutables, barreras que multiplican nuestra importancia y deciden qué tanto valemos."* Son las matrices de parámetros entrenables que multiplican a la entrada y al estado anterior respectivamente, regulando qué tanta importancia e influencia tiene cada componente en el cálculo.

### El vector de sesgo o bias ($b$)
La frase asociada es: *"un pequeño desvío inevitable"*. Representa el sesgo o bias que se suma a la combinación lineal para desplazar la función de activación, permitiendo que la red se ajuste mejor a los datos sin estar atada a pasar rígidamente por el origen del plano.

### La función de activación tangente hiperbólica ($\tanh$)
La frase asociada es: *"un muro curvo que nos comprime entre el -1 y el 1, evitando que nuestra energía explote hacia el infinito."* Es la función de activación no lineal que escala los valores para mantenerlos en un rango controlado entre menos uno y uno, lo que evita que los números crezcan indefinidamente durante los cálculos recurrentes.

### El estado oculto actual ($h_t$)
La frase asociada es: *"una nueva identidad. Soy tu estado actual, la respuesta de hoy, y estoy listo para ser el fantasma de tu mañana."* Representa el nuevo estado de memoria de la red, que funciona como la salida en este paso de tiempo y a la vez se convierte en el estado del pasado ($h_{t-1}$) para la siguiente iteración.

## 3. Actividad 2.2: Análisis de Dimensionalidad (El tamaño de los peajes)

Supongamos que estamos diseñando esta red para procesar secuencias de datos donde el vector de entrada actual ($x_t$) tiene una dimensión de 20 y el espacio del estado oculto anterior ($h_{t-1}$) tiene una dimensión de 64. A continuación se detallan los tamaños requeridos para cada matriz de pesos y la justificación matemática de cada una.

### Dimensiones de la matriz de entrada ($W_{hx}$)
La matriz de pesos $W_{hx}$ debe tener un tamaño de $64 \times 20$. Matemáticamente esto se justifica porque debe multiplicar al vector de entrada $x_t$, que mide $20 \times 1$. Al hacer el producto matricial de una matriz de tamaño $64 \times 20$ por un vector columna de $20 \times 1$, el resultado es un vector columna de tamaño $64 \times 1$, el cual ya se puede sumar directamente con los otros componentes del espacio oculto que miden 64.

### Dimensiones de la matriz recurrente ($W_{hh}$)
La matriz recurrente $W_{hh}$ debe medir $64 \times 64$. La justificación matemática es que esta matriz multiplica al vector del estado oculto anterior $h_{t-1}$, que tiene un tamaño de $64 \times 1$. Para mantener la información dentro del mismo espacio oculto de 64 características, necesitamos una transformación lineal cuadrada de $64 \times 64$ que dé como resultado un nuevo vector columna de $64 \times 1$.

### Dimensión final de la salida ($h_t$)
La dimensión final del vector resultante $h_t$ es de 64 (un vector columna de tamaño $64 \times 1$). Esto ocurre porque $h_t$ es el resultado de aplicar la función de activación $\tanh$ elemento por elemento a la suma de tres vectores que miden $64 \times 1$: el término de la entrada ($W_{hx} x_t$), el término del estado recurrente ($W_{hh} h_{t-1}$) y el vector de sesgo ($b$, que también debe medir $64 \times 1$).

```python
import numpy as np

# Definir dimensiones de la entrada y del espacio oculto
dim_entrada = 20
dim_oculto = 64

# Simular vectores de entrada (x_t) y estado anterior (h_t_minus_1)
x_t = np.random.randn(dim_entrada, 1)      # Vector columna 20x1
h_t_minus_1 = np.random.randn(dim_oculto, 1)  # Vector columna 64x1

# Inicializar matrices de pesos con las dimensiones calculadas
W_hx = np.random.randn(dim_oculto, dim_entrada)  # Matriz de pesos 64x20
W_hh = np.random.randn(dim_oculto, dim_oculto)   # Matriz recurrente 64x64
b = np.random.randn(dim_oculto, 1)               # Vector de sesgo 64x1

# Realizar multiplicaciones y sumas lineales
termino_entrada = np.dot(W_hx, x_t)
termino_recurrente = np.dot(W_hh, h_t_minus_1)
combinacion_lineal = termino_entrada + termino_recurrente + b

# Aplicar la función de activación tangente hiperbólica
h_t = np.tanh(combinacion_lineal)

# Imprimir las dimensiones para verificar los cálculos
print("=== VERIFICACIÓN DE DIMENSIONES EN PYTHON ===")
print(f"Dimensión de entrada x_t: {x_t.shape}")
print(f"Dimensión del estado oculto anterior h_(t-1): {h_t_minus_1.shape}")
print(f"Dimensión de la matriz W_hx: {W_hx.shape}")
print(f"Dimensión de la matriz recurrente W_hh: {W_hh.shape}")
print(f"Dimensión del sesgo b: {b.shape}")
print(f"Dimensión del término W_hx * x_t: {termino_entrada.shape}")
print(f"Dimensión del término W_hh * h_(t-1): {termino_recurrente.shape}")
print(f"Dimensión del estado oculto resultante h_t: {h_t.shape}")
```
