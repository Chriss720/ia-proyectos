### Preguntas de análisis
1. **¿La fila de COME se parece a la de FRUTA? ¿Por qué deberían diferir?**
   * No. Tienen que ser distintas porque son tipos de palabras diferentes. `COME` (verbo) busca quién realiza la acción (`NIÑA`) y qué se come (`FRUTA`), mientras que `FRUTA` (objeto) busca principalmente la acción que la afecta (`COME`).
2. **¿Alguna fila reparte atención casi pareja (todas ~20 %)? ¿Qué palabra podría ser esa y por qué?**
   * La fila de **NIÑA** es la más repartida. Al ser el sustantivo principal, se conecta con casi todas las palabras de la oración (su artículo `LA`, su adjetivo `PEQUEÑA`, el verbo `COME` y sí misma).
3. **Si la frase tuviera 100 palabras, ¿cuántas celdas tendría la tabla? ¿Por qué eso explica que textos muy largos cuestan más memoria?**
   * Tendría **10,000 celdas** (100x100). Esto muestra que la memoria crece de forma cuadrática ($N^2$). Entre más largo el texto, la matriz de atención se vuelve gigante y consume muchísimos recursos.

## Actividad 10 — Atención cruzada (decoder mirando encoder)

* **Encoder (Español):** YO QUIERO CAFE
* **Decoder (Inglés):** I WANT ___ (Generando la tercera palabra: coffee)

### Paso 1: Matriz rectangular 3x3 (Fila del hueco 3)

| Desde (inglés) ↓ / Español → | YO | QUIERO | CAFE |
| --- | :---: | :---: | :---: |
| **Palabra 3 (por escribir)** | 1 | 2 | 10 |

### Paso 2: Porcentajes y respuestas
* **Porcentajes de la fila 3 (Suma = 13):**
  * **YO**: 7.7%
  * **QUIERO**: 15.4%
  * **CAFE**: 76.9%

1. **¿CAFE debería ganar? ¿Por qué?**
   * Sí, porque la palabra que se está por escribir en inglés es "coffee", que traduce directamente "CAFE".
2. **¿La fila de I (primera palabra generada) podría mirar mucho a YO? ¿Tiene sentido?**
   * Sí, porque "I" se traduce como "YO". Tiene sentido para alinear los pronombres.
3. **Diferencia clave:**
   * En **self-attention** las filas y las columnas son del mismo texto/idioma. En **cross-attention**, las filas son del decoder (idioma destino) y las columnas del encoder (idioma origen).

## Actividad 13 — RNN vs Transformer: contar conexiones

Nodos: **A — B — C — D — E**

### Respuestas
1. **Si A debe influir en E, ¿cuántos “saltos” en RNN? ¿Y en una capa de atención?**
   * En RNN: **4 saltos** (de uno en uno: A $\to$ B $\to$ C $\to$ D $\to$ E).
   * En atención: **1 solo salto** (cualquier palabra puede mirar directamente a otra).
2. **¿Cuál crece más rápido si la frase tiene 100 palabras: 100 enlaces secuenciales o celdas?**
   * Las **celdas** crecen muchísimo más rápido. La RNN tiene $N-1$ enlaces (99), mientras que la atención tiene $N^2$ celdas (10,000).
3. **¿Por qué aun así usamos Transformers y no solo RNN?**
   * Porque permiten calcular toda la frase al mismo tiempo en paralelo (más rápido) y no pierden memoria en frases largas, ya que conectan palabras lejanas directamente.

## Actividad 14 — Escalar (evitar saturación)

### Ejercicios de Softmax aproximado

1. **Softmax aproximado sin escalar [8, 2, 2, 2]:**
   * Exponenciales: $e^8 \approx 2980.96$, $e^2 \approx 7.39$
   * Suma: $2980.96 + 7.39 \times 3 = 3003.13$
   * **Probabilidades:**
     * P1: **99.3%**
     * P2, P3, P4: **0.2%** cada una
     * *Resultado:* El valor alto domina casi al 100% (saturación).

2. **Softmax aproximado con escalar [4, 1, 1, 1] (dividido entre 2):**
   * Exponenciales: $e^4 \approx 54.60$, $e^1 \approx 2.72$
   * Suma: $54.60 + 2.72 \times 3 = 62.76$
   * **Probabilidades:**
     * P1: **87.0%**
     * P2, P3, P4: **4.3%** cada una

3. **Comparación:**
   * La palabra ganadora sigue ganando, pero ahora tiene un porcentaje menor (de 99.3% bajó a 87.0%). Esto evita la saturación extrema y permite que las otras palabras sigan aportando información a la red.
